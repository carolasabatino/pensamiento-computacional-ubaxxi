```python
# Ejercicio 2

entero_1 = int(input("Escriba un entero: "))
entero_2 = int(input("Escriba otro entero: "))

suma = entero_1 + entero_2
resta = entero_1 - entero_2
multiplicacion = entero_1 * entero_2
division = entero_1 / entero_2
resto = entero_1 % entero_2

print("Suma: ", suma)
print("Resta: ", resta)
print("Multiplicacion: ", multiplicacion)
print("Division: ", division)
print("Resto: ", resto)
```
